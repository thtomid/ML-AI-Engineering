```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import pandas as pd
from numpy import dtypes

torch.manual_seed(42)
```

```python
predictions = torch.tensor([[6.0],[3.0]], dtype=torch.float)
targets = torch.tensor([[4.0],[5.0]], dtype=torch.float)

raw_errors = predictions - targets
squared_error = raw_errors ** 2
manual_mse = squared_error.mean()
```

```python
mse_loss_function = nn.MSELoss()
pytorch_mse = mse_loss_function(predictions, targets)
pytorch_mse
```

```python
print("Predictions: ")
print(predictions.squeeze())
print("Targets: ")
print(targets.squeeze())
print("Raw errors: ")
print(raw_errors.squeeze())
print("Squared error: ")
print(squared_error.squeeze())
print("manual mse: ")
print(manual_mse)
print("Pytorch mse: ")
print(pytorch_mse)
```

```python
cross_entropy_loss_function = nn.CrossEntropyLoss()

logits = torch.tensor([[1.0,-1.0,2.0,-0.5,0.0,-0.8,0.1,4.0,0.2,-0.3]], dtype=torch.float)
labels = torch.tensor([3], dtype=torch.long)

loss = cross_entropy_loss_function(logits, labels)
probabilities = torch.softmax(logits, dim=1)
predicted_class = probabilities.argmax(dim=1)
```

```python
print("Logits shape: ", logits.shape)
print("Labels shape: ", labels.shape)
print("Correct: ", labels.item())
print("Predicted class: ", predicted_class.item())
```
