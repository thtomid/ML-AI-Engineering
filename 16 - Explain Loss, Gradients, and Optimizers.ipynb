```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import pandas as pd
from numpy import dtypes

torch.manual_seed(42)
```

```python
predictions = torch.tensor([[6.0],[3.0]], dtype=torch.float)
targets = torch.tensor([[4.0],[5.0]], dtype=torch.float)

raw_errors = predictions - targets
squared_error = raw_errors ** 2
manual_mse = squared_error.mean()
```

```python
mse_loss_function = nn.MSELoss()
pytorch_mse = mse_loss_function(predictions, targets)
pytorch_mse
```

```python
print("Predictions: ")
print(predictions.squeeze())
print("Targets: ")
print(targets.squeeze())
print("Raw errors: ")
print(raw_errors.squeeze())
print("Squared error: ")
print(squared_error.squeeze())
print("manual mse: ")
print(manual_mse)
print("Pytorch mse: ")
print(pytorch_mse)
```
