```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
study_hours = np.array([1,2,3,4,5,6,7], dtype = float)
quiz_score = np.array([56,61,67,72,78,83,89], dtype = float)

df = pd.DataFrame({
    "study_hours": study_hours,
    "quiz_score": quiz_score
})
```

```python
df
```

```python
plt.figure(figsize=(8,5))
plt.scatter(study_hours, quiz_score)
plt.xlabel('Study Hours')
plt.ylabel('Quiz score')
plt.title('Data Points Plot')
plt.grid(True)
plt.show()
```

```python
def line(x, weight, bias):
    return weight * x + bias
```

```python
bad_weight = 2.0
bad_bias = 50.0

x_line = np.linspace(study_hours.min(), study_hours.max(), 100)
y_bad = line(x_line, bad_weight, bad_bias)

plt.figure(figsize=(8,5))
plt.scatter(study_hours, quiz_score)
plt.plot(x_line, y_bad)
plt.xlabel('Study Hours')
plt.ylabel('Quiz score')
plt.title(f'Bad Fit: weight={bad_weight}, bias={bad_bias}')
plt.grid(True)
plt.show()
```
